#Repeat first slice of a manually fixed atlas

```python
path = Path(R"SPIRIT dev 20261005\processed\SPIRIT dev 20261006_GRADIENT_FFE_Thermometry_11_1_atlas_manual_fix.nii.gz")
```

```python
nii = nib.load(path)
data = nii.get_fdata()
data = np.repeat(data[..., [0]], data.shape[-1], axis = -1)
nii = nib.Nifti1Image(data, nii.affine, header = nii.header)
nib.save(nii, path)
```
